# Data Scraping

Before we can predict electricity prices, we need raw data describing the price signal itself and the conditions around it. We collect four datasets: day-ahead prices, electricity consumption, observed weather, and historical weather forecasts.

The scraping functions all follow the same pattern: they call the relevant API, save a raw CSV file in `data/`, and return the result as a pandas DataFrame.

| Data | Why we need it | Source | Output file |
|------|----------------|--------|-------------|
| Day-ahead prices | The price series we want to predict | Energi Data Service | `data/day_ahead_prices_dk1_raw.csv` |
| Electricity consumption | Describes system load in DK1 | Energi Data Service | `data/consumption_dk1_raw.csv` |
| Weather actuals | Historical weather observations | Open-Meteo Archive API | `data/weather_actuals_raw.csv` |
| Weather forecasts | Forecasts as they were known before each target hour | Open-Meteo Previous Runs API | `data/weather_forecasts_raw.csv` |

The datasets can be fetched one at a time. This is useful when only one raw file needs to be updated.

```python
from src.data.data_collection import (
    fetch_day_ahead_prices,
    fetch_consumption,
    fetch_weather_actuals,
    fetch_weather_forecasts,
)

prices = fetch_day_ahead_prices(
    start="2021-01-01",
    end="2026-04-28",
    price_area="DK1",
)

consumption = fetch_consumption(
    start="2021-01-01",
    end="2026-04-28",
    price_area="DK1",
)

weather_actuals = fetch_weather_actuals(
    start="2021-01-01",
    end="2026-04-28",
)

weather_forecasts = fetch_weather_forecasts(
    start="2025-01-01",
    end="2026-04-28",
)
```

For the full project setup, we usually collect everything through `fetch_all()`. This runs the four calls above and returns the results in a dictionary.

```python
from src.data.data_collection import fetch_all

results = fetch_all(
    start="2021-01-01",
    end="2026-04-28",
    price_area="DK1",
)
```
